```python

gamma = 0.9
theta = 0.001

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

policy = {
    'S0': 'Up',
    'S1': 'Up'
}

V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S0', -1),
        'Right': ('S0', -1)
    },
    'S1': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S1', -1),
        'Right': ('G', 100)
    }
}

def policy_evaluation():
    while True:
        delta = 0

        for s in ['S0', 'S1']:
            old_value = V[s]

            next_state, reward = transitions[s][policy[s]]

            V[s] = reward + gamma * V[next_state]

            delta = max(delta, abs(old_value - V[s]))

        if delta < theta:
            break

def policy_improvement():
    stable = True

    for s in ['S0', 'S1']:
        old_action = policy[s]

        q_values = {}

        for a in actions:
            next_state, reward = transitions[s][a]
            q_values[a] = reward + gamma * V[next_state]

        best_action = max(q_values, key=q_values.get)

        policy[s] = best_action

        if old_action != best_action:
            stable = False

        print(s, "Q-values:", q_values)

    return stable

iteration = 0

while True:
    iteration += 1

    policy_evaluation()

    print("\nIteration:", iteration)
    print("State Values:", V)

    if policy_improvement():
        break

print("\nFinal Optimal State Values:")
for s in states:
    print(s, "=", V[s])

print("\nOptimal Policy:")
for s in ['S0', 'S1']:
    print(s, "=", policy[s])

print("\nBellman Optimality Verification:")

for s in ['S0', 'S1']:
    q_values = []

    for a in actions:
        next_state, reward = transitions[s][a]
        q = reward + gamma * V[next_state]
        q_values.append(q)

    print(s, "V* =", V[s], "Max Q* =", max(q_values))
```
