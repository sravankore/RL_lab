```python
#ASSIGNMENT-1
n = int(input("Enter number of time steps: "))
rewards = []
for i in range(n):
    reward = float(input(f"Enter reward for time step {i}: "))
    rewards.append(reward)
undiscounted_return = sum(rewards)
gamma = float(input("Enter discount factor (gamma): "))
discounted_return = 0
for i in range(n):
    discounted_return += (gamma ** i) * rewards[i]
print("\nEntered Reward Values:", rewards)
print("Undiscounted Return:", undiscounted_return)
print("Discounted Return:", discounted_return)
```

```python
#ASSIGNMENT-2
STATES = ["S0", "S1", "S3", "S4", "G"]
ACTIONS = ["Up", "Down", "Left", "Right"]
REWARDS = {
    "S0": -1,
    "S1": -1,
    "S3": -1,
    "S4": -1,
    "G": 10
}
GAMMA = 0.9
TRANSITIONS = {
    "S0": {
        "Up": "S0",
        "Down": "S3",
        "Left": "S0",
        "Right": "S1"
    },

    "S1": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S0",
        "Right": "G"
    },

    "S3": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S3",
        "Right": "S4"
    },

    "S4": {
        "Up": "G",
        "Down": "S4",
        "Left": "S3",
        "Right": "S4"
    },

    "G": {
        "Up": "G",
        "Down": "G",
        "Left": "G",
        "Right": "G"
    }
}
def get_next_state(state, action):
    return TRANSITIONS[state][action]
```

```python
# main.py
#ASSIGNMENT-2
print("2 x 3 GRID WORLD")
print("----------------")
print("State Space:", STATES)
print("Action Space:", ACTIONS)
print("Discount Factor:", GAMMA)
state = input("\nEnter current state: ")
action = input("Enter action (Up/Down/Left/Right): ")
if state in STATES and action in ACTIONS:
    next_state = get_next_state(state, action)
    print("\nCurrent State:", state)
    print("Action:", action)
    print("Next State:", next_state)
    if next_state == "G":
        print("Reward: +10")
    else:
        print("Reward: -1")

else:
    print("Invalid state or action")
```
