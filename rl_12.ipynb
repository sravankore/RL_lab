```python

gamma = 0.9
theta = 0.001

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S0', -1),
        'Right': ('S0', -1)
    },
    'S1': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S1', -1),
        'Right': ('G', 100)
    }
}

iteration = 0

while True:
    delta = 0
    new_V = V.copy()

    for s in ['S0', 'S1']:
        q_values = {}

        for a in actions:
            next_state, reward = transitions[s][a]
            q_values[a] = reward + gamma * V[next_state]

        new_V[s] = max(q_values.values())

        delta = max(delta, abs(new_V[s] - V[s]))

    V = new_V
    iteration += 1

    print("\nIteration:", iteration)
    print("S0 =", round(V['S0'], 3))
    print("S1 =", round(V['S1'], 3))

    if delta < theta:
        break

print("\nFinal Optimal State Values:")

for s in states:
    print(s, "=", round(V[s], 3))

print("\nFinal Action Values and Optimal Policy:")

for s in ['S0', 'S1']:
    q_values = {}

    for a in actions:
        next_state, reward = transitions[s][a]
        q_values[a] = reward + gamma * V[next_state]

    best_action = max(q_values, key=q_values.get)

    print("\nState:", s)
    print("Q-values:", q_values)
    print("Optimal Action:", best_action)
```
