```python

gamma = 0.9
theta = 0.0001

states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']
goal = 'G'

transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S0', -1),
        'Right': ('S0', -1)
    },
    'S1': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S1', -1),
        'Right': ('G', 10)
    }
}

print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)

print("\nTransition Table:")
for s in states:
    for a in actions:
        ns, r = transitions[s][a]
        print(s, a, ns, r)

V = {'S0': 0, 'S1': 0, 'G': 0}

print("\nInitial Values:", V)

iteration = 0

while True:
    iteration += 1
    new_V = V.copy()

    for s in states:
        q_values = []

        for a in actions:
            ns, reward = transitions[s][a]
            q = reward + gamma * V[ns]
            q_values.append(q)

        new_V[s] = max(q_values)

    delta = max(abs(new_V[s] - V[s]) for s in states)

    V = new_V

    print("Iteration", iteration, V, "Delta =", delta)

    if delta < theta:
        break

print("\nFinal Optimal State Values:")
for s in states:
    print(s, "=", V[s])

print("\nFinal Action Values:")
Q = {}

for s in states:
    Q[s] = {}

    for a in actions:
        ns, reward = transitions[s][a]
        Q[s][a] = reward + gamma * V[ns]

    print(s, Q[s])

print("\nOptimal Policy:")

for s in states:
    best_value = max(Q[s].values())
    best_actions = [a for a in actions if abs(Q[s][a] - best_value) < 1e-9]
    print(s, "=", best_actions)

print("\nBellman Optimality Verification:")

for s in states:
    print(s, "V* =", V[s],
          "Max Q =", max(Q[s].values()))

print("\nOptimal Path: S0 -> S1 -> G")
```
