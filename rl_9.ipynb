```python

gamma = 0.9

A = 1 - (0.9 * 0.9 * 0.9)

V0 = (-1 - 0.9 * 1) / A
V1 = -1 + 0.9 * V0

Q_S0_Down = -1 + gamma * V1
Q_S0_Right = -1

Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V0

print("State Values:")
print("V(S0) =", round(V0, 2))
print("V(S1) =", round(V1, 2))

print("\nAction Values:")
print("Q(S0, Down) =", round(Q_S0_Down, 2))
print("Q(S0, Right) =", round(Q_S0_Right, 2))
print("Q(S1, Right) =", round(Q_S1_Right, 2))
print("Q(S1, Up) =", round(Q_S1_Up, 2))

print("\nVerification:")
print("V(S0) =", 0.9 * Q_S0_Down + 0.1 * Q_S0_Right)
print("V(S1) =", 0.9 * Q_S1_Right + 0.1 * Q_S1_Up)
```
