```python
import math

# Machine names
machines = ['A', 'B', 'C']

# 10 plays
selected_machines = ['A', 'B', 'C', 'B', 'B', 'B', 'A', 'C', 'A', 'C']

# Rewards according to the selected machines
rewards = [3, 6, 4, 7, 5, 8, 4, 3, 2, 5]

# Initialize
N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]
UCB = [0.0, 0.0, 0.0]

# Perform 10 plays
for play in range(10):

    machine = selected_machines[play]
    reward = rewards[play]

    # Find machine index
    i = machines.index(machine)

    # Update count
    N[i] += 1

    # Update total reward
    R[i] += reward

    # Incremental sample-average Q-value
    Q[i] = Q[i] + (reward - Q[i]) / N[i]

    # Calculate UCB
    for j in range(3):
        if N[j] == 0:
            UCB[j] = float('inf')
        else:
            UCB[j] = Q[j] + math.sqrt(
                (2 * math.log(play + 1)) / N[j]
            )

# Display results
print("====================================")
print("       FINAL RESULTS")
print("====================================")

print("Machine\tN\tR\tQ\tUCB")

for i in range(3):
    if UCB[i] == float('inf'):
        ucb = "Infinity"
    else:
        ucb = round(UCB[i], 4)

    print(
        machines[i],
        "\t",
        N[i],
        "\t",
        R[i],
        "\t",
        round(Q[i], 2),
        "\t",
        ucb
    )

# Total reward
total_reward = sum(R)

print("====================================")
print("Total Reward =", total_reward)
print("Total Plays  =", sum(N))

# Best machine
best_machine = machines[Q.index(max(Q))]

print("Best Machine =", best_machine)
print("Best Q-value =", max(Q))
print("====================================")
```
