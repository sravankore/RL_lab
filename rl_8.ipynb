```python
# grid_env_det.py

STATES = ["S0", "S1", "S3", "S4", "G"]

ACTIONS = ["Up", "Down", "Left", "Right"]

GAMMA = 0.9

REWARDS = {
    "S0": -1,
    "S1": -1,
    "S3": -1,
    "S4": -1,
    "G": 10
}

TRANSITIONS = {
    "S0": {
        "Up": "S0",
        "Down": "S3",
        "Left": "S0",
        "Right": "S1"
    },

    "S1": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S0",
        "Right": "G"
    },

    "S3": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S3",
        "Right": "S4"
    },

    "S4": {
        "Up": "G",
        "Down": "S4",
        "Left": "S3",
        "Right": "S4"
    },

    "G": {
        "Up": "G",
        "Down": "G",
        "Left": "G",
        "Right": "G"
    }
}


def get_next_state(state, action):
    return TRANSITIONS[state][action]


policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}

def state_value_function(state):

    current_state = state
    discounted_return = 0
    discount = 1

    while current_state != "G":

        action = policy[current_state]

        next_state = get_next_state(
            current_state, action
        )

        reward = REWARDS[next_state]

        discounted_return += discount * reward

        discount *= GAMMA

        current_state = next_state

    return discounted_return

def action_value_function(state, action):

    next_state = get_next_state(state, action)

    reward = REWARDS[next_state]

    if next_state == "G":
        return reward

    value_after_action = state_value_function(next_state)

    return reward + GAMMA * value_after_action

print("2 x 3 GRID WORLD")
print("----------------")

print("Policy:")
print("S0 -> Right")
print("S1 -> Down")
print("S3 -> Right")
print("S4 -> Up")

state_value = state_value_function("S0")

action_value = action_value_function("S0", "Up")

print("\nState Value of S0:", state_value)
print("Action Value of S0, Up:", action_value)
```
