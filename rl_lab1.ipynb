```python
n = int(input("Enter number of machines: "))

q = {}

for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards: ").split()))

    q[name] = sum(rewards) / len(rewards)

    print("Machine", name)
    print("Rewards:", rewards)
    print("Q(" + name + ") =", q[name])

best = max(q, key=q.get)

print("Greedy Selected Machine:", best)
```

```python
n = int(input("Enter number of machines: "))

Q = {}
N = {}
rewards = {}

for i in range(n):
    name = input("Enter machine name: ")
    r = list(map(int, input("Enter rewards: ").split()))

    rewards[name] = r
    Q[name] = 0
    N[name] = 0

print("\nUpdating Q values...\n")

for name in rewards:
    for R in rewards[name]:
        N[name] += 1
        Q[name] = Q[name] + (R - Q[name]) / N[name]

        print("Machine:", name)
        print("Reward:", R)
        print("N =", N[name])
        print("Q =", Q[name])
        print()

print("Final Q Values:")

for name in Q:
    print("Q(" + name + ") =", Q[name])

best = max(Q, key=Q.get)

print("\nGreedy Selected Machine:", best)
```
